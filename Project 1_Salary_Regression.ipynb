```python
# importing relevant frameworks
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
# load datasets 
df=pd.read_csv(r"C:\Users\user\Downloads\Salary_Data.csv")
df.head()
```

```python
# Initial Data Analysis
df.dtypes
df.columns
df.shape
df.info()
```

```python
# Finding missing values
print(df.isna().sum())
```

```python
df.describe()
```

```python
# determining features and variables
x = df[["YearsExperience"]]
y = df[["Salary"]]
```

```python
# splitting dataset
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)

```

```python
# training the model
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# model predictions
y_predict=model.predict(x_test)
print(y_predict)
```

```python
# Calculating MSE
mse=mean_squared_error(y_test,y_predict)
print(mse)
```

```python
# finding the intercept and slope for the regression line
print(model.intercept_)
print(model.coef_[0])
```

```python
# visualizing the regression line
plt.scatter(x,y,color="purple",label="Actual Data")
plt.plot(x,model.predict(x),color="blue",linewidth=2,label="Regression Line")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Salary Predictions based on Years of Experience")
plt.legend()
plt.show()
```
